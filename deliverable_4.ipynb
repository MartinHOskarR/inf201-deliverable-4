## Task 2: Fractions

The `Fraction` class stores an integer numerator and a positive integer denominator. Comparisons use cross-multiplication: a/b < c/d is the same as a·d < c·b when b and d are positive. Besides the initialiser and the six comparisons, it has `__add__`, `__sub__`, `__mul__` and `__str__`.

```python
from fraction import Fraction

a = Fraction(1, 2)
b = Fraction(1, 3)
print(a + b, a - b, a * b)
print(a > b)
```
